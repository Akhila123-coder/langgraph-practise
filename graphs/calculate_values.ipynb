```python
from langgraph.graph import StateGraph
from typing import List, TypedDict
```

```python
from typing import List, TypedDict, Optional

class AgentState(TypedDict):
  name: str
  values: List[int]
  operator: str
  result: Optional[str]
```

```python
def caluculate(state: AgentState) -> AgentState:
  if state["operator"] == "+":
    res = sum(state["values"])
  elif state["operator"] == "*":
    res = 1
    for val in state["values"]:
      res *= val
  state["result"] = f"Hi {state['name']}, here if the result of this operator ({state["operator"]}) = {res}"
  return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("caluculator", caluculate)

graph.set_entry_point('caluculator')
graph.set_finish_point('caluculator')

app = graph.compile()
```

```python
res = app.invoke({"name":"Akhila", "values": [1, 2, 3, 4, 5], "operator": "*"})

res['result']
```

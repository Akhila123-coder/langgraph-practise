```python
from langgraph.graph import StateGraph
from typing import Dict, TypedDict
```

```python
class AgentState(TypedDict):
    message: str

def say_thanks(state: AgentState):
    state['message'] = "Hi " + state["message"] + ", thank you for helping!" 
    return state

```

```python
graph = StateGraph(AgentState)

graph.add_node("thanker", say_thanks)

graph.set_entry_point("thanker")
graph.set_finish_point("thanker")

app = graph.compile()
```

```python
from IPython.display import Image, display
display(Image(app.get_graph().draw_mermaid_png()))
```

```python
app.invoke({"message": "Sara"})
```
